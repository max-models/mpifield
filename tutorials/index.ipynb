# Tutorials

Jupyter notebooks, executed when the documentation is built. Download a notebook from its page to run it yourself.

- [Getting started](/mpifield/tutorials/getting-started/): create, fill, evaluate, plot and combine fields.
- [Gather and scatter](/mpifield/tutorials/gather-and-scatter/): deposit markers and interpolate at them.
- [MPI and halos](/mpifield/tutorials/mpi-and-halos/): fields split over ranks, deposits across block edges.
- [The kernels](/mpifield/tutorials/kernels/): the compiled kernels, their versions, and calling them directly.
