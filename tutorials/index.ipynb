# Examples

Jupyter notebooks, executed when the documentation is built. Download a notebook from its page to run it yourself.

- [Getting started](/mpifield/examples/getting-started/): create, fill, evaluate, plot and combine fields.
- [Gather and scatter](/mpifield/examples/gather-and-scatter/): deposit markers and interpolate at them.
- [MPI and halos](/mpifield/examples/mpi-and-halos/): fields split over ranks, deposits across block edges.
- [The kernels](/mpifield/examples/kernels/): the compiled kernels, their versions, and calling them directly.
- [Scripts](/mpifield/example-scripts/) are complete Python scripts with their plots.
